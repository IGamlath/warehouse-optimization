## Directory Guidelines

- **`data/raw/`** — Store original, unmodified datasets downloaded from external sources.
- **`data/processed/`** — Store cleaned, transformed, and analysis-ready datasets.
- **`notebooks/`** — Perform the project's data analysis, visualization, modeling, and simulation in Jupyter notebooks.
- **`src/`** — Store reusable Python modules if needed; the primary analysis remains notebook-based.
- **`reports/`** — Store final project reports, methodology, and results summaries.
- **`figures/`** — Store exported charts, graphs, and other visualizations.

## Project Conventions

1. Preserve original source files in `data/raw/`.
2. Use project-relative paths instead of hardcoded absolute paths.
3. Document assumptions, transformations, and modeling decisions in the notebooks.
4. Keep real observations distinct from synthetic or simulated data.
5. Exclude the virtual environment, large datasets, and sensitive credentials from Git commits.
6. Create each notebook as it becomes necessary rather than creating every planned notebook upfront.

**Note:** This structure is the intended layout. Some folders and notebooks may not exist yet.
